```python
import numpy as np
from pixell import enmap, curvedsky, utils

# ------------------------------------------------------------------
# 1. Paramètres généraux de la carte
# ------------------------------------------------------------------
# Taille angulaire du patch (10° x 10°)
size_deg = 10.0

# Résolution (en arcmin) - suffisante pour atteindre ℓ_max ≃ 7950
res_arcmin = 0.5
res_rad = res_arcmin * utils.arcmin  # utils.arcmin = π / (180*60)

# Nombre de pixels par côté : (10° * 60') / 0.5' = 1200
npix = int(round(size_deg * 60.0 / res_arcmin))

# Géométrie pixell (projection CAR, carte centrée autour de (ra,dec)=(0,0))
# Géométrie pixell (projection CAR)
shape, wcs = enmap.geometry(
    pos=(0.0 * utils.degree, 0.0 * utils.degree),   # centre RA=0°, Dec=0°
    shape=(npix, npix),
    res=res_rad,
    proj="car"
)

print("shape =", shape, "npix =", npix, "res_arcmin =", res_arcmin)

# ------------------------------------------------------------------
# 2. Chargement de la courbe de bruit SO baseline tSZ
# ------------------------------------------------------------------
# Fichier SO (tSZ, baseline, post-component-separation, f_sky=0.4)
# -> N_ell^{yy} en unités de Compton-y^2, sans facteur ℓ(ℓ+1)/2π
so_noise_file = "/chemin/vers/so_noise_models/LAT_comp_sep_noise/v3.1.0/" \
                "SO_LAT_Nell_T_atmv1_baseline_fsky0p4_ILC_tSZ.txt"

ell_tab, Nell_tab = np.loadtxt(so_noise_file, unpack=True)

# ell_tab : multipôles (ℓ)
# Nell_tab : N_ell^{yy} (non beam-deconvolved, en Compton-y^2)

# ------------------------------------------------------------------
# 3. Construction du spectre N_ell sur tous les ℓ, + fenêtre tanh 80<ℓ<7950
# ------------------------------------------------------------------
ell_max = int(ell_tab.max())
ells = np.arange(ell_max + 1)

# Interpolation linéaire (en ℓ) des valeurs N_ell
# (les fichiers SO ne sont pas forcément définis pour tous les entiers ℓ)
Nell = np.interp(ells, ell_tab, Nell_tab)

# Fenêtre tanh comme dans Alina : on coupe ℓ ≤ 80 et ℓ ≥ 7950 en douceur
lmin, lmax = 80, 7950
delta = 10.0  # largeur de la transition tanh ~ [80,90] et [7940,7950]

def tanh_window(ell, lmin, lmax, delta):
    """
    Fenêtre lisse ~1 pour lmin+δ/2 < ℓ < lmax-δ/2,
    qui tombe vers 0 en dehors, avec transitions tanh
    dans [lmin, lmin+δ] et [lmax-δ, lmax].
    """
    # montée basse-ℓ
    x_low = (ell - (lmin + 0.5 * delta)) / (0.5 * delta)
    f_low = 0.5 * (1.0 + np.tanh(x_low))
    # descente haute-ℓ
    x_high = ((lmax - 0.5 * delta) - ell) / (0.5 * delta)
    f_high = 0.5 * (1.0 + np.tanh(x_high))
    return f_low * f_high

window = tanh_window(ells, lmin, lmax, delta)
Cl_noise = Nell * window

# ℓ=0,1 sans intérêt pour un patch plat -> on force à 0
Cl_noise[:2] = 0.0

# ------------------------------------------------------------------
# 4. Préparation pour curvedsky.rand_map
# ------------------------------------------------------------------
# ps doit être de taille (ncomp, ncomp, ℓ_max+1)
ps = np.zeros((1, 1, Cl_noise.size))
ps[0, 0] = Cl_noise  # champ scalaire unique avec spectre N_ell^{yy}

# ------------------------------------------------------------------
# 5. Génération d'une réalisation de bruit SO baseline tSZ
# ------------------------------------------------------------------
# La carte retournée est en unités de Compton-y, avec le N_ell
# non-déconvolu, donc déjà au beam effectif du modèle SO.
noise_map = curvedsky.rand_map(
    shape, wcs, ps,
    lmax=ell_max,
    seed=None   # mettre un entier pour reproduire toujours la même réalisation
)[0]

print("Carte de bruit générée, min/max =", noise_map.min(), noise_map.max())

# ------------------------------------------------------------------
# 6. Sauvegarde en FITS
# ------------------------------------------------------------------
out_file = "SO_baseline_ILC_tSZ_noise_10x10deg_0p5arcmin.fits"
enmap.write_map(out_file, noise_map)
print("Carte écrite dans", out_file)
```

```python
import optuna
import os

# 1. Configuration (Doit être identique au script d'entraînement)
STUDY_NAME = "optimize_npe_wst"
# Le dossier où se trouve le fichier .db généré par le script précédent
STORAGE_DIR = "/rds/rds-clecat/pipeline_alina_full/alina_paper/optuna_studies"
STORAGE_URL = f"sqlite:///{os.path.join(STORAGE_DIR, STUDY_NAME)}.db"

# 2. Chargement de l'étude
print(f"Chargement de l'étude depuis {STORAGE_URL}...")
study = optuna.load_study(study_name=STUDY_NAME, storage=STORAGE_URL)

print(f"Nombre d'essais réalisés : {len(study.trials)}")
print(f"Meilleure Loss trouvée : {study.best_value}")
print("Meilleurs paramètres :")
for key, val in study.best_params.items():
    print(f"  - {key}: {val}")

# --- VISUALISATIONS ---

# 3. Historique d'optimisation
# Permet de voir si l'algorithme converge (si les points descendent vers le bas à droite)
fig1 = optuna.visualization.plot_optimization_history(study)
fig1.show()

# 4. Importance des hyperparamètres
# C'est LE graphique le plus utile. Il te dit quel paramètre influence le plus le résultat.
# Exemple : Si "initial_lr" est à 0.8, c'est que le learning rate est critique.
# Si "hidden_features" est très bas, c'est que la taille du réseau importe peu.
fig2 = optuna.visualization.plot_param_importances(study)
fig2.show()

# 5. Graphique de contour (Slice Plot)
# Montre la distribution de la Loss pour chaque paramètre individuellement.
# Utile pour voir les "bons coins". Par exemple, tu verras peut-être que
# pour le Learning Rate, tout ce qui est au-dessus de 1e-3 donne de mauvais résultats.
fig3 = optuna.visualization.plot_slice(study)
fig3.show()

# 6. Parallel Coordinate Plot
# Utile pour voir les corrélations. Par exemple :
# "Les modèles qui marchent bien ont souvent un petit Learning Rate ET un grand Batch Size"
fig4 = optuna.visualization.plot_parallel_coordinate(study)
fig4.show()
```

```python
# Au lieu de fig1.show(), utilise :
fig1.write_html("optuna_history.html")
fig2.write_html("optuna_importance.html")
fig3.write_html("optuna_slice.html")
fig4.write_html("optuna_parallel.html")

print("Graphiques sauvegardés en fichiers HTML dans le dossier courant.")
```
